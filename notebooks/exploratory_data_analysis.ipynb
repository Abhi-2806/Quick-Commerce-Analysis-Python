```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
import plotly.express as px
```

```python
# Load Dataset

df = pd.read_csv (r"C:\Users\a0808\OneDrive\Desktop\Q-Comm\quick_commerce_data_raw.csv")
```

```python
#Data Structure

df.info()

#Summary Statistics

df.describe()


# Shape of the dataset
df.shape


# Check Missing Values NULL & NA

df.isna().sum()


# To show the records which contains atleat ANY one or ALL null values.

df[df.isna().any(axis=1)]
```

```python
# Drop those rows that have missing values in those column which are of STRING/CATEGORY Format 
#eg; City is a string format hence we can't use mean/mode method to replace those missing values

df= df.dropna(subset= ['City'] )
```

```python
# Check Central Tendency

mode = df['Items_Count'].mode()
median = df['Items_Count'].median()
mean = df['Items_Count'].mean()
```

```python
#Fill the missing value with MODE
#[0] = index value 0
#inplace = to make the changes permanent

df['Items_Count'].fillna(mode[0], inplace = True)
```

```python
# Fill Null values with Group-wise Mean/Median (To fill blank customer_rating column)

df['Customer_Rating'] = df.groupby("Company") ["Customer_Rating"].transform(lambda x:x.fillna(x.mean() ))
```

```python
# Fill remaining null Values of customer_rating column with mean value

df['Delivery_Partner_Rating'] = df['Delivery_Partner_Rating'].fillna (df['Delivery_Partner_Rating'].mean())
```

```python
# Cross-Check if all the missing values had been replaced or not .!

df.isna().sum()
```

## Removing Outliers

```python
# Checking the Outliers

# Remove Order_ID column bcoz we have all unique id upto 1 million, therefore we will drop that particular column

plt.figure(figsize= (15,6))

df.drop(columns = ["Order_ID"]).boxplot()
```

```python
# Removing Outlier [ FILTERING METHOD ]

df = df[df['Order_Value'] <= 2500]
```

```python
# Plotting a box plot to represent the outliers

plt.figure(figsize= (15,6))

df.drop(columns = ["Order_ID"]).boxplot(vert = False , showmeans = True )
```

## Changing the Data Type of the columns

```python
#Dataset Structure

df .info()
```

```python
# Change the data type of Order ID 

df['Order_ID'] = df['Order_ID'].astype(str)
```

```python
# Round-off the float values

df['Order_Value']= np.round(df['Order_Value'])
df['Delivery_Time_Min']= np.round(df['Delivery_Time_Min'])
df['Distance_Km']= np.round(df['Distance_Km'], 1)
df['Customer_Rating']= np.round(df['Customer_Rating'])
df['Delivery_Partner_Rating']= np.round(df['Delivery_Partner_Rating'])
```

```python
# Change the data-type of float columns to integer

df['Order_Value']= df ['Order_Value'] .astype(int)
df['Delivery_Time_Min']= df['Delivery_Time_Min']. astype(int)
df['Items_Count']= df['Items_Count']. astype(int)
df['Customer_Rating']= df['Customer_Rating']. astype(int)
df['Delivery_Partner_Rating']= df['Delivery_Partner_Rating']. astype(int)
```
