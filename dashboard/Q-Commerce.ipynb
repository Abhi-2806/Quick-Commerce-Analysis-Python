## Q8) which company has the best operational efficiency (Delivery Time vs Order Volume)?

```python
df.head ()
```

```python
# Group data and apply aggregate functions

company_efficiency = df.groupby('Company').agg(
                           Total_Orders = ('Order_ID', 'count' ) ,
                            Avg_Delivery_Time = ('Delivery_Time_Min' , 'mean' )
                            ).reset_index()

company_efficiency
```

```python
# Normalize the value for fair comparision as order value is very high as compared to delivery time

from sklearn.preprocessing import MinMaxScaler

# Scaling both the column to the same range (0 to 1) so that they can be fairly  compared and combined.

scaler = MinMaxScaler()  # create a scaling object

company_efficiency [['Total_Orders_Scaled' , 'Avg_Delivery_Time_Scaled']] = scaler.fit_transform (
                                      company_efficiency [['Total_Orders' , 'Avg_Delivery_Time' ]]          
                                            
                                            )

    
```

```python
company_efficiency
```

```python
# Efficiency score (hiher orders, lower time)

company_efficiency['Efficiency Score'] = company_efficiency['Total_Orders_Scaled'] - company_efficiency['Avg_Delivery_Time_Scaled']
```

```python
company_efficiency
```

```python
# Sort by best efficiency

Eff_Sort = company_efficiency.sort_values('Efficiency Score', ascending = False)

Eff_Sort
```

```python
x = company_efficiency['Efficiency Score'].unique()

x
```

```python
color = sns.color_palette("coolwarm", len(x))
```

```python
plt.bar(Eff_Sort['Company'], Eff_Sort['Efficiency Score'], color = color , edgecolor = 'black')

plt.xlabel('Company')
plt.ylabel('Efficiency Score')
plt.title('Operational Efficiency by Company')

plt.grid(True)

plt.xticks(rotation = 30, ha ='right')

plt.tight_layout()
plt.show()
```

```python
# Use plotly library - Scatter Bubble plot

import plotly.express as px

# Create positive bubble size
min_eff = Eff_Sort['Efficiency Score'].min()
Eff_Sort['Bubble_Size'] = Eff_Sort['Efficiency Score'] - min_eff + 0.1


# PLOT
fig = px.scatter(Eff_Sort, x = 'Total_Orders' ,  y = 'Avg_Delivery_Time' , size ='Bubble_Size', color = 'Company', text = 'Company' ,
                title = 'Operational Efficiency of Quick Commerce Platforms',
                hover_data = {'Total_Orders' : True , 'Avg_Delivery_Time': ':.2f' , 'Efficiency Score' : ':.3f'  },
                size_max = 40)


# Improve chart
fig.update_layout (
                    height = 650,
                    width = 1100,
                    xaxis_title = 'Total Orders (Higher is better)',
                    yaxis_title = 'Average delivery timee (Lower is better)',
                    title_x = 0.5,
                    template = "plotly_dark"
                    )
# Make labels cleaner
fig.update_traces(textposition = 'middle center')

# Reverse Y-axis so best performance appers higher
fig.update_yaxes(autorange = 'reversed' )

fig.show()
```

### Alternative KPI Chart creation

```python
# ==========================================
# QUICK COMMERCE KPI DASHBOARD
# ==========================================

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.ticker import FuncFormatter

# ------------------------------------------
# 1. CREATE COMPANY-LEVEL KPI TABLE
# ------------------------------------------

company_metrics = (
    df.groupby("Company")
      .agg(
          Orders=("Order_ID", "count"),
          Revenue=("Order_Value", "sum"),
          Avg_Delivery_Time=("Delivery_Time_Min", "mean"),
          Avg_Rating=("Customer_Rating", "mean")
      )
      .reset_index()
)

# Round decimal values
company_metrics["Avg_Delivery_Time"] = (
    company_metrics["Avg_Delivery_Time"].round(1)
)

company_metrics["Avg_Rating"] = (
    company_metrics["Avg_Rating"].round(2)
)


# ------------------------------------------
# 2. DASHBOARD STYLE
# ------------------------------------------

sns.set_theme(style="whitegrid")

plt.rcParams["font.family"] = "sans-serif"
plt.rcParams["font.size"] = 11

fig, axes = plt.subplots(
    nrows=2,
    ncols=2,
    figsize=(18, 12)
)

# Overall dashboard background
fig.patch.set_facecolor("#F4F6F7")


# ------------------------------------------
# 3. COMMON CHART FORMATTING FUNCTION
# ------------------------------------------

def style_chart(ax, title, xlabel):
    ax.set_title(
        title,
        fontsize=16,
        fontweight="bold",
        color="#1B2631",
        pad=15
    )

    ax.set_xlabel(
        xlabel,
        fontsize=11,
        fontweight="bold"
    )

    ax.set_ylabel("")

    # Keep only horizontal value-axis gridlines
    ax.grid(axis="x", linestyle="--", alpha=0.30)
    ax.grid(axis="y", visible=False)

    # Remove extra chart borders
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)
    ax.spines["left"].set_visible(False)

    # Individual chart card background
    ax.set_facecolor("white")


# ------------------------------------------
# 4. ORDERS BY COMPANY
# ------------------------------------------

orders_data = company_metrics.sort_values(
    by="Orders",
    ascending=True
)

ax1 = axes[0, 0]

sns.barplot(
    data=orders_data,
    x="Orders",
    y="Company",
    palette="Blues_d",
    ci=None,
    ax=ax1
)

style_chart(
    ax1,
    "Orders by Company",
    "Number of Orders"
)

# Add labels manually
orders_max = orders_data["Orders"].max()

for bar in ax1.patches:
    value = bar.get_width()

    ax1.text(
        value + (orders_max * 0.015),
        bar.get_y() + bar.get_height() / 2,
        f"{value:,.0f}",
        va="center",
        fontsize=10,
        fontweight="bold",
        color="#1B2631"
    )

# Add space at the end for labels
ax1.set_xlim(0, orders_max * 1.18)


# ------------------------------------------
# 5. REVENUE BY COMPANY
# ------------------------------------------

revenue_data = company_metrics.sort_values(
    by="Revenue",
    ascending=True
)

ax2 = axes[0, 1]

sns.barplot(
    data=revenue_data,
    x="Revenue",
    y="Company",
    palette="YlOrBr",
    ci=None,
    ax=ax2
)

style_chart(
    ax2,
    "Revenue by Company",
    "Revenue"
)

# Revenue axis format
ax2.xaxis.set_major_formatter(
    FuncFormatter(lambda x, pos: f"{x / 1000000:.1f}M")
)

# Add revenue labels manually
revenue_max = revenue_data["Revenue"].max()

for bar in ax2.patches:
    value = bar.get_width()

    ax2.text(
        value + (revenue_max * 0.015),
        bar.get_y() + bar.get_height() / 2,
        f"{value / 1000000:.2f}M",
        va="center",
        fontsize=10,
        fontweight="bold",
        color="#1B2631"
    )

ax2.set_xlim(0, revenue_max * 1.22)


# ------------------------------------------
# 6. AVERAGE DELIVERY TIME BY COMPANY
# ------------------------------------------

delivery_data = company_metrics.sort_values(
    by="Avg_Delivery_Time",
    ascending=True
)

ax3 = axes[1, 0]

sns.barplot(
    data=delivery_data,
    x="Avg_Delivery_Time",
    y="Company",
    palette="magma",
    ci=None,
    ax=ax3
)

style_chart(
    ax3,
    "Average Delivery Time by Company",
    "Delivery Time (Minutes)"
)

# Add delivery-time labels manually
delivery_max = delivery_data["Avg_Delivery_Time"].max()

for bar in ax3.patches:
    value = bar.get_width()

    ax3.text(
        value + (delivery_max * 0.02),
        bar.get_y() + bar.get_height() / 2,
        f"{value:.1f} min",
        va="center",
        fontsize=10,
        fontweight="bold",
        color="#1B2631"
    )

ax3.set_xlim(0, delivery_max * 1.25)


# ------------------------------------------
# 7. AVERAGE CUSTOMER RATING BY COMPANY
# ------------------------------------------

rating_data = company_metrics.sort_values(
    by="Avg_Rating",
    ascending=True
)

ax4 = axes[1, 1]

sns.barplot(
    data=rating_data,
    x="Avg_Rating",
    y="Company",
    palette="viridis",
    ci=None,
    ax=ax4
)

style_chart(
    ax4,
    "Average Customer Rating by Company",
    "Customer Rating (Out of 5)"
)

# Ratings are usually between 0 and 5
ax4.set_xlim(0, 5.8)

# Add rating labels manually
for bar in ax4.patches:
    value = bar.get_width()

    ax4.text(
        value + 0.06,
        bar.get_y() + bar.get_height() / 2,
        f"{value:.2f}",
        va="center",
        fontsize=10,
        fontweight="bold",
        color="#1B2631"
    )


# ------------------------------------------
# 8. MAIN DASHBOARD TITLE
# ------------------------------------------

fig.suptitle(
    "Quick Commerce KPI Dashboard",
    fontsize=25,
    fontweight="bold",
    color="#0B1F5E",
    y=0.98
)

fig.text(
    0.5,
    0.945,
    "Company-Level Performance Overview",
    ha="center",
    fontsize=13,
    color="#566573"
)

# Adjust spacing so titles and labels never overlap
plt.tight_layout(rect=[0, 0, 1, 0.90])

plt.show()
```
